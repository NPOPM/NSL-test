```python
'''
张量变形
'''
x = torch.randn(2, 4, 8)  #batch = 2, seq_len = 4, dim = 8
```

```python
#view:改变形状，但数据在内存中必须连续
x_view = x.view(2, 2, 4, 4)  #拆成2个batch，每个里面两个头
print(x_view.shape)  #[2, 2, 4, 4]
```

```python
#transpose:交换两个维度
x_t = x.transpose(1, 2)  #把第一维和第二维交换
print(x_t.shape)  #[2, 8, 4]
```

```python
#permute:任意重排维度顺序
x_p = x.permute(0, 2, 1)  #把[batch, seq, dim]变成[batch, dim, seq]
print(x_p.shape)  #[2, 8, 4]
```

```python
#做完transpose或permute后，内存不连续了，view会报错，这时候需要先.contiguous()
x_t_contig = x.transpose(1, 2).contiguous()
x_t_contig.view(2, -1, 8)  #-1表示自动推算
print(x_t_contig.shape)
```

```python
'''
矩阵乘法
'''
q = torch.randn(2, 4, 8)  #[batch, seq_len, dim]
k = torch.randn(2, 4, 8)
v = torch.randn(2, 4, 8)
```

```python
#方式1:matmul，最通用
scores = torch.matmul(q, k.transpose(1, 2))  #[2,4,4]
print(scores.shape)
```

```python
#方式2:@运算符，和matmul等价
scores2 = q @ k.transpose(1, 2)
print(scores2.shape)
```

```python
#方式3:bmm(batch矩阵乘法)，要求输入是3维
scores3 = torch.bmm(q, k.transpose(1, 2))
print(scores3.shape)
```

```python
'''
Softmax(注意力权重归一化)
'''
x = torch.randn(2,4,4)
```

```python
#对最后一维做softmax
probs = torch.softmax(x, dim=-1)#dim=-1表示最后一维
print(probs.shape)#[2,4,4]
print(probs[0,0].sum())
```

```python
'''
拼接和拆分
'''
#cat:沿某个维度拼接
a=torch.randn(2,3,4)
b=torch.randn(2,5,4)
c=torch.cat([a,b],dim=1)#沿第一维拼接->[2,8,4]
print(c.shape)
```

```python
#split:按大小拆分
x=torch.randn(2,8,4)
chunks=x.split(2,dim=1)#沿第一维每两个拆一份->四个[2,2,4]的块
print(chunks[0].shape)
```

```python
#chunk:按数量拆分
chunks = x.chunk(4,dim=1)#沿第一维拆成四份2->四个[2,2,4]的块
print(chunks[0].shape)
```
